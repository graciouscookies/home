```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("melb_data.csv")
```

```python
print("shape: ",df.shape)
print("missing values: ",df.isnull().sum())
print("data types: ",df.dtypes)
print("zeroes: ",(df==0).sum())
print("shape: ",df['Price'].mean())
print("statistics: ",df.describe())
```

```python
df['Car']=df['Car'].fillna(df['Car'].mean())
df['BuildingArea']=df['BuildingArea'].fillna(df['BuildingArea'].mean())
df['YearBuilt']=df['YearBuilt'].fillna(df['YearBuilt'].mean())
df['CouncilArea']=df['CouncilArea'].fillna(df['CouncilArea'].mode()[0])
```

```python
df['Price'].hist(bins=40)
plt.xlabel("Price")
plt.ylabel("number of properties")
plt.title("price vs number")
plt.show()
```
