```python
from queue import PriorityQueue

class Node:
    def __init__(self,state,parent,g,h):
        self.parent=parent
        self.state=state
        self.g=g
        self.h=h
        self.f=g+h

    def __lt__(self,other):
        return self.state<other.state

def print_path(node,show=None):
    path=[]
    while node:
        path.append(node.state)
        node=node.parent
    path.reverse()
    if show:
        for s in show:
            show(s)
    else:
        print("->".join(str(s) for s in path))

def astar(start,goal,get_neighbours,heuristic,show=None):
    open_list=PriorityQueue()
    open_list.put(Node(start,None,0,heuristic(start,goal)))
    best_g={start:0}
    while not open_list.empty():
        current=open_list.get()
        if current.state==goal:
            print("goal reached.total cost: ",current.g)
            return
        if current.g>best_g[current.state]:
            continue
        for next_state, cost in get_neighbours(current.state):
            g=current.g+cost
            if next_state not in best_g or g<best_g[next_state]:
                best_g[next_state]=g
                open_list.put(Node(next_state,current,g,heuristic(next_state,goal)))
    print("no solution")
```

```python
def puzzle_neighbours(s):
    result=[]
    i=s.index(0)
    r,c=divmod(i,3)
    for dr,dc in [(-1,0),(0,-1),(1,0),(0,1)]:
        nr,nc=r+dr,c+dc
        if 0<=nr<3 and 0<=nc<3:
            j=nr*3+nc
            t=list(s)
            t[i],t[j]=t[j],t[i]
            result.append((tuple(t),1))
    return result

def puzzle_h(s,goal):
    d=0
    for i,v in enumerate(s):
        if v!=0:
            g=goal.index(v)
            d+=abs(i//3-g//3)+abs(i%3-g%3)
    return d

def show_board(s):
    for i in (0,3,6):
        print(s[i:i+3])
    print()

start=(1,2,3,6,4,5,7,0,8)
astar(start,(1,2,3,4,5,6,7,8,0),puzzle_neighbours,puzzle_h,show_board)
```

```python
grid=[[0,0,0,0],
    [1,1,0,1],
    [0,0,0,0],
    [0,1,1,0]]

def grid_neighbours(p):
    r,c=p
    result=[]
    for dr,dc in [(-1,0),(0,-1),(1,0),(0,1)]:
        nr,nc=r+dr,c+dc
        if 0<=nr<len(grid) and 0<=nc<len(grid[0]) and grid[nr][nc]==0:
            result.append(((nr,nc),1))
    return result

def grid_h(p,goal):
    return abs(p[0]-goal[0])+abs(p[1]-goal[1])

astar((0,0),(3,3),grid_neighbours,grid_h)
```
