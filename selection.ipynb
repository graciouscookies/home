```python
def selection_sort(arr):
    arr=arr.copy()
    n=len(arr)
    for i in range(n-1):
        m=i
        for j in range(i+1,n):
            if arr[j]<arr[m]:
                m=j
        arr[i],arr[m]=arr[m],arr[i]
        print(f"pass {i+1} : minimum : {arr[i]} , array : {arr}")
    print("final array",arr)
    return arr

selection_sort([16,42,3,23,19,80])
```
