```python
import heapq

def dijikstras(graph,src):
    dist={v: float('inf') for v in graph }
    prev={v: None for v in graph}
    dist[src]=0
    pq=[(0,src)]
    while pq:
        d,u=heapq.heappop(pq)
        if d>dist[u]:
            continue
        for v,w in graph[u]:
            if w+d<dist[v]:
                dist[v]=d+w
                prev[v]=u
                heapq.heappush(pq,(dist[v],v))

    print(f"Minimum path from source {src}: ")
    for v in graph:
        path,x=[],v
        while x is not None:
            path.append(x)
            x=prev[x]
        print(f"{src}->{v} : Distance: {dist[v]} , Path : {'->'.join(map(str,path[::-1]))}")
    return dist,prev

graph={'A':[('B',4),('C',4)],
       'B':[('C',4),('D',2)],
       'C':[('A',1),('B',3)],
       'D':[('C',4),('A',1)]}

dijikstras(graph,'A')
```
