```python
class Disjoinset:
    def __init__(self,vertices):
        self.parent={v:v for v in vertices}

    def find(self,x):
        if self.parent[x]!=x:
            self.parent[x]=self.find(x)
        return self.parent[x]

    def union(self,x,y):
        self.parent[self.find(y)]=self.find(x)

def kruskals(edges):
    vertices={v for u,v,w in edges} | {u for u,v,w in edges}
    ds=Disjoinset(vertices)
    edges=sorted(edges,key=lambda e:e[2])
    print("sorted edges: ",edges)
    mst,total=[],0
    for u,v,w in edges:
        if ds.find(u)==ds.find(v):
            print(f"{u}-{v} ({w}) SAME cycle rejected")
        else:
            ds.union(u,v)
            mst.append((u,v,w))
            total+=w
            print(f"{u}-{v} ({w}) diff cycle accepted")
        if len(mst)<len(vertices)-1:
            break
    if len(mst)<len(vertices)-1:
        print("disjoint")
    print("cost: ",total)
    print("mst: ",mst)

edges = [('A', 'B', 2), ('A', 'C', 3), ('B', 'C', 1), ('B', 'D', 1),
         ('C', 'D', 4), ('C', 'E', 5), ('D', 'E', 2)]
kruskals(edges)
```
