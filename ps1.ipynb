```python
import glob
import pandas as pd
from sklearn.model_selection import train_test_split
import kagglehub
```

```python
path = kagglehub.dataset_download("zhaoyingzhu/heartcsv")
df=pd.read_csv(glob.glob(path+"/*.csv")[0])
```

```python
print("shape: ",df.shape)
print("missing values: ",df.isnull().sum())
print("data types: ",df.dtypes)
print("zeroes: ",(df==0).sum())
print("mean age: ",df['Age'].mean())
```

```python
df2=df[['Age', 'Sex', 'ChestPain', 'RestBP', 'Chol']]
train, test = train_test_split(df2, test_size=0.25, random_state=42)
print("Train shape: ",train.shape,"Test shape: ",test.shape)
```
