```python
import heapq

def prims(graph,start):
    visited={start}
    pq=[]
    for v,w in graph[start]:
        heapq.heappush(pq,(w,start,v))
    mst,total=[],0
    while pq and len(visited)<len(graph):
        w,u,v=heapq.heappop(pq)
        if v in visited:
            continue
        visited.add(v)
        mst.append((u,v,w))
        total+=w
        print(f"step{len(mst)} from {u} - {v} : weight {w}")
        for x,wx in graph[v]:
            if x not in visited:
                heapq.heappush(pq,(wx,v,x))

    if len(visited)!=len(graph):
        print("disconnected")
    print("mst: ",mst)
    print("cost: ",total)

graph = {
    'A': [('B', 2), ('C', 3)],
    'B': [('A', 2), ('C', 1), ('D', 1)],
    'C': [('A', 3), ('B', 1), ('D', 4), ('E', 5)],
    'D': [('B', 1), ('C', 4), ('E', 2)],
    'E': [('C', 5), ('D', 2)],
}

prims(graph,'A')
```
